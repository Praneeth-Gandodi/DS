```python
import pandas as pd
```

```python
index = pd.MultiIndex.from_tuples(
    [
        ('Engineering', 'CSE'),
        ('Engineering', 'ECE'),
        ('Science', 'Physics'),
        ('Science', 'Chemistry')
    ],
    names=['Department', 'Branch']
)
```

```python
print(index)
```

```python
data = pd.DataFrame(
    {
        '2025': [85, 78, 92, 88],
        '2026': [90, 82, 95, 91]
    },
    index=index
)
```

```python
print(data)
```

```python
unstacked_data = data.unstack() 
```

```python
print(unstacked_data)
```

```python
stacked_data = unstacked_data.stack() 
```

```python
print(stacked_data) 
```
